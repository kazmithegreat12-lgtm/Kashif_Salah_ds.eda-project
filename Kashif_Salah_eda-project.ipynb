```python
import matplotlib.patches as mpatches
import matplotlib.pyplot as plt
import seaborn as sns

# Set clean aesthetic style
sns.set_theme(style="whitegrid")
plt.figure(figsize=(9, 5))

# 1. Filter dataset for Nicole's budget ($350k–$550k) and target bedrooms (2, 3, 4)
nicole_df = df[(df["price"] >= 350000) & (df["price"] <= 550000)].copy()
filtered_beds = nicole_df[nicole_df["bedrooms"].isin([2, 3, 4])]

# 2. Calculate median living space and median price per bedroom count
summary = (
    filtered_beds.groupby("bedrooms")
    .agg(
        median_sqft=("sqft_living", "median"),
        median_price=("price", "median"),
    )
    .reset_index()
)

# 3. Create vertical bar plot
x_labels = ["2 Bedrooms", "3 Bedrooms", "4 Bedrooms"]
colors = ["#90CAF9", "#4A90E2", "#1B60A5"]

bars = plt.bar(
    x_labels,
    summary["median_sqft"],
    color=colors,
    width=0.45,
    edgecolor="white",
    linewidth=1.2,
)

# 4. Add exact Square Footage AND Price on top of each bar
for i, bar in enumerate(bars):
    sqft = summary.loc[i, "median_sqft"]
    price = summary.loc[i, "median_price"]

    plt.text(
        bar.get_x() + bar.get_width() / 2.0,
        sqft + 50,
        f"{sqft:,.0f} sq ft\n(Median: ${price/1000:.0f}k)",
        ha="center",
        va="bottom",
        fontsize=10,
        fontweight="bold",
        color="#0D47A1",
    )

# 5. Add a clear legend for stakeholders
patch_2bed = mpatches.Patch(color="#90CAF9", label="2 Beds: Starter / Compact")
patch_3bed = mpatches.Patch(
    color="#4A90E2", label="3 Beds: Optimal Value & Space"
)
patch_4bed = mpatches.Patch(color="#1B60A5", label="4 Beds: Larger Footprint")

plt.legend(
    handles=[patch_2bed, patch_3bed, patch_4bed],
    title="Property Category",
    loc="upper left",
    frameon=True,
    facecolor="white",
    fontsize=9.5,
)

# 6. Formatting titles and axes
plt.title(
    "Hypothesis 2: Living Space & Price by Bedroom Count ($350k–$550k)",
    fontsize=13,
    weight="bold",
    pad=15,
)
plt.xlabel("Bedroom Configuration", fontsize=11, labelpad=10)
plt.ylabel("Median Living Area (Sq Ft)", fontsize=11, labelpad=10)

# Set Y-axis limit to give padding for callout text
plt.ylim(0, summary["median_sqft"].max() * 1.3)

sns.despine()
plt.tight_layout()
plt.show()
```

Code and Vizualization for our 3. Hypothesis

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# --------------------------------------------------
# 1. Filter houses within Nicole's target budget
# --------------------------------------------------

df_budget = df[
    (df["price"] >= 350000) &
    (df["price"] <= 550000)
].copy()

# Remove rows with missing values in relevant columns
df_budget = df_budget.dropna(
    subset=["price", "zipcode", "lat", "long"]
)

# --------------------------------------------------
# 2. Define Central Seattle
# --------------------------------------------------

center_lat = 47.6062
center_long = -122.3321

# --------------------------------------------------
# 3. Calculate distance to Central Seattle
#    using the Haversine formula
# --------------------------------------------------

lat1 = np.radians(center_lat)
lon1 = np.radians(center_long)

lat2 = np.radians(df_budget["lat"])
lon2 = np.radians(df_budget["long"])

dlat = lat2 - lat1
dlon = lon2 - lon1

a = (
    np.sin(dlat / 2) ** 2
    + np.cos(lat1)
    * np.cos(lat2)
    * np.sin(dlon / 2) ** 2
)

df_budget["dist_to_center_km"] = (
    6371 * 2 * np.arcsin(np.sqrt(a))
)

# --------------------------------------------------
# 4. Summarize by ZIP code
# --------------------------------------------------

zip_summary = (
    df_budget
    .groupby("zipcode")
    .agg(
        num_homes=("price", "count"),
        median_dist=("dist_to_center_km", "median")
    )
    .reset_index()
)

# --------------------------------------------------
# 5. Define reference values
# --------------------------------------------------

median_distance = zip_summary["median_dist"].median()
median_inventory = zip_summary["num_homes"].median()

# --------------------------------------------------
# 6. Define ideal area:
#    closer than median + more homes than median
# --------------------------------------------------

ideal_zone = zip_summary[
    (zip_summary["median_dist"] <= median_distance) &
    (zip_summary["num_homes"] >= median_inventory)
].copy()

# Top 5 ZIP codes:
# highest availability first, then shorter distance
top5 = (
    ideal_zone
    .sort_values(
        by=["num_homes", "median_dist"],
        ascending=[False, True]
    )
    .head(5)
)

# --------------------------------------------------
# 7. Create scatter plot
# --------------------------------------------------

plt.figure(figsize=(10, 6))

# All ZIP codes in background
plt.scatter(
    zip_summary["median_dist"],
    zip_summary["num_homes"],
    color="lightgrey",
    s=60,
    alpha=0.6,
    label="Other ZIP codes"
)

# Highlight Top 5
plt.scatter(
    top5["median_dist"],
    top5["num_homes"],
    color="teal",
    s=130,
    alpha=0.9,
    edgecolors="black",
    label="Top 5 ZIP codes"
)

# --------------------------------------------------
# 8. Label Top 5 without overlap
# --------------------------------------------------

top5_labels = top5.sort_values(
    "num_homes",
    ascending=False
).reset_index(drop=True)

label_offsets = [
    (12, 18),
    (12, 5),
    (12, -12),
    (-50, 18),
    (-50, -12)
]

for i, row in top5_labels.iterrows():

    offset = label_offsets[i]

    plt.annotate(
        str(int(row["zipcode"])),
        (row["median_dist"], row["num_homes"]),
        xytext=offset,
        textcoords="offset points",
        fontsize=9,
        fontweight="bold",
        bbox=dict(
            boxstyle="round,pad=0.25",
            facecolor="white",
            edgecolor="grey",
            alpha=0.85
        ),
        arrowprops=dict(
            arrowstyle="-",
            color="grey",
            alpha=0.6
        )
    )

# --------------------------------------------------
# 9. Add reference lines
# --------------------------------------------------

plt.axvline(
    median_distance,
    linestyle="--",
    color="grey",
    alpha=0.6,
    label="Median distance"
)

plt.axhline(
    median_inventory,
    linestyle="--",
    color="grey",
    alpha=0.6,
    label="Median availability"
)

# --------------------------------------------------
# 10. Highlight ideal top-left area
# --------------------------------------------------

y_max = zip_summary["num_homes"].max() * 1.12

plt.fill_between(
    [0, median_distance],
    median_inventory,
    y_max,
    color="green",
    alpha=0.04,
    label="Closer + more choices"
)

# --------------------------------------------------
# 11. Titles and labels
# --------------------------------------------------

plt.title(
    "Best ZIP Codes for Nicole: Budget vs. Distance",
    fontsize=13
)

plt.xlabel(
    "Median Distance to Central Seattle (km)",
    fontsize=10
)

plt.ylabel(
    "Number of Homes within $350k–$550k",
    fontsize=10
)

plt.ylim(0, y_max)

plt.grid(
    True,
    linestyle="--",
    alpha=0.2
)

plt.legend(
    loc="upper right"
)

plt.tight_layout()
plt.show()
```

We are using Scatter Plot for our Hypothesis_3.

This scatter plot maps housing availability within Nicole's budget against the median distance to central Seattle. The shaded green area highlights the optimal quadrant—ZIP codes that offer both above-average home inventory and a below-average commute. The top 5 recommended ZIP codes (like 98034 and 98133) stand out at the top of this zone for delivering the highest concentration of options.
